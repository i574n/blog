#### rss (test)

```spiral
:= item (base : string) (e : page.entry) : string =
    := link = base ++# "/" ++# html.escape e.slug ++# "/"
    "<item><title>"
    ++# html.escape e.title
    ++# "</title><link>"
    ++# link
    ++# "</link><guid>"
    ++# link
    ++# "</guid>"
    ++# "<dc:date>"
    ++# html.escape e.date
    ++# "</dc:date><description>"
    ++# html.escape e.blurb
    ++# "</description></item>"

:= ^ items (base : string) (entries : list page.entry) : string =
    ?= entries ?|
    | [] => ""
    | e :: rest => item base e ++# items base rest

:= render_feed (title : string) (base : string) (entries : list page.entry) : string =
    "<?xml version=\"1.0\" encoding=\"UTF-8\"?>"
    ++# "<rss version=\"2.0\" xmlns:dc=\"http://purl.org/dc/elements/1.1/\"><channel>"
    ++# "<title>"
    ++# html.escape title
    ++# "</title><link>"
    ++# base
    ++# "/</link><description>"
    ++# html.escape title
    ++# "</description>"
    ++# items base entries
    ++# "</channel></rss>"
```

#### the_feed_links_each_post_by_its_slug (test)

```spiral
///- --test --symbol
///> zig

render_feed "Spiral Blog" "https://i574n.github.io/blog" [ page.sample () ]
|> sm'.contains "<item><title>Sample Entry</title><link>https://i574n.github.io/blog/sample/</link>"
|> _assert_eq !+
```

#### the_feed_is_rss_2_with_dublin_core_dates (test)

```spiral
///- --test --symbol
///> zig

render_feed "Spiral Blog" "https://i574n.github.io/blog" [ page.sample () ]
|> \ feed =>
    (feed |> sm'.starts_with "<?xml version=\"1.0\" encoding=\"UTF-8\"?><rss version=\"2.0\"")
    && (feed |> sm'.contains "<dc:date>2026-10-17</dc:date>")
|> _assert_eq !+
```
