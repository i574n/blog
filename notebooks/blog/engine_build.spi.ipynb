```spiral
:= ^ write_files (root : string) (files : list site.file) : () =
    ?= files ?|
    | [] => ()
    | f :: rest =>
        := target = root </> f.path
        target
        |> file_system.directory_get_parent
        |> optionm'.default_value' root
        |> file_system.create_dir_all
        |> resultm.unwrap'
        f.content |> file_system.write_all_text target
        write_files root rest
```
