```spiral
:= escape (text : string) : string =
    text
    |> sm'.replace "&" "&amp;"
    |> sm'.replace "<" "&lt;"
    |> sm'.replace ">" "&gt;"
    |> sm'.replace "\"" "&quot;"
```

#### escape_keeps_markup_out_of_text (test)

```spiral
///- --test

escape "<a href=\"x\"> & b" |> _assert_eq "&lt;a href=&quot;x&quot;&gt; &amp; b"
```

#### escape_leaves_plain_text_alone (test)

```spiral
///- --test

escape "Behind the Scenes" |> _assert_eq "Behind the Scenes"
```
